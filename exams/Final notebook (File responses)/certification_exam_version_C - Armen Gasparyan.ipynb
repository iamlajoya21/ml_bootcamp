**Task:** Complete the function below so that the visible checks pass. / **Առաջադրանք՝** Լրացրե՛ք ֆունկցիան, որպեսզի տեսանելի թեստերը անցնեն։

**Q94 — your code / ձեր կոդը**

> Write your solution in the code cell below. / Գրե՛ք ձեր լուծումը ներքևի code cell-ում։

````python
import numpy as np

def standardize(x):
    x_arr = np.array(x, dtype=float)
    mean = np.mean(x_arr)
    std = np.std(x_arr)  # numpy.std-ն հաշվում է պոպուլյացիոն ստանդարտ շեղումը (ddof=0)
    
    if std == 0:
        return np.zeros_like(x_arr)
    
    return (x_arr - mean) / std
````

````python
import numpy as np
x = np.array([-2, 0, 2, 4], dtype=float)
out = standardize(x)
assert out.shape == x.shape
assert np.allclose(out.mean(), 0.0)
assert np.allclose(out.std(), 1.0)
print('visible checks passed')
````

### Q95 — Pandas — hard

**EN.** Implement `clean_and_summarize(df)`: copy the frame, parse `timestamp` as datetime, fill missing `score` with the mean of its `team`, create `scaled` = `score` * 10, and return the mean of `scaled` per `team` sorted by group.

**HY.** Իրականացրե՛ք `clean_and_summarize(df)`-ը․ պատճենե՛ք frame-ը, `timestamp`-ը վերածե՛ք datetime-ի, լրացրե՛ք բացակայող `score`-ն իր `team`-ի միջինով, ստեղծե՛ք `scaled` = `score` * 10 և վերադարձրե՛ք `scaled`-ի միջինը յուրաքանչյուր `team`-ի համար՝ դասավորված ըստ group-ի։

**Task:** Complete the function below so that the visible checks pass. / **Առաջադրանք՝** Լրացրե՛ք ֆունկցիան, որպեսզի տեսանելի թեստերը անցնեն։

**Q95 — your code / ձեր կոդը**

> Write your solution in the code cell below. / Գրե՛ք ձեր լուծումը ներքևի code cell-ում։

````python
import pandas as pd

def clean_and_summarize(df):
    
    df_copy = df.copy()
    
    df_copy['timestamp'] = pd.to_datetime(df_copy['timestamp'])
    
    df_copy['score'] = df_copy['score'].fillna(df_copy.groupby('team')['score'].transform('mean'))
    
    df_copy['scaled'] = df_copy['score'] * 10
    
    result = df_copy.groupby('team')['scaled'].mean().sort_index()
    
    return result
````

**EN.** Implement `forecast_next(series)` using lag-1 and lag-2 features with a linear regression. Return the one-step-ahead prediction as a float.

**HY.** Իրականացրե՛ք `forecast_next(series)`-ը՝ օգտագործելով lag-1 և lag-2 հատկանիշներ և գծային regression։ Վերադարձրե՛ք մեկ քայլ առաջ կանխատեսումը որպես float։

**Task:** Complete the function below so that the visible checks pass. / **Առաջադրանք՝** Լրացրե՛ք ֆունկցիան, որպեսզի տեսանելի թեստերը անցնեն։

**Q98 — your code / ձեր կոդը**

> Write your solution in the code cell below. / Գրե՛ք ձեր լուծումը ներքևի code cell-ում։

````python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression

def forecast_next(series):
    
    s = pd.Series(series)

    df = pd.DataFrame()
    df['y'] = s
    df['lag_1'] = s.shift(1)
    df['lag_2'] = s.shift(2)
    
    df = df.dropna()
    
    X = df[['lag_1', 'lag_2']]
    y = df['y']
    
    model = LinearRegression()
    model.fit(X, y)
    
    last_lag_1 = s.iloc[-1]
    last_lag_2 = s.iloc[-2]
    
    X_next = np.array([[last_lag_1, last_lag_2]])
    
    prediction = model.predict(X_next)[0]
    
    return float(prediction)
````

````python
import numpy as np
series = [10, 20, 30, 40, 50]
pred = forecast_next(series)
assert isinstance(pred, float)
assert abs(pred - 60) < 0.5, pred
print('visible checks passed', round(pred, 3))
````

### Q99 — Recommenders — hard

**EN.** Implement `recommend_items(matrix, user_idx, k)`: compute item-item cosine similarity, take the user's highest-rated item, and return the indices of the k most similar other items.

**HY.** Իրականացրե՛ք `recommend_items(matrix, user_idx, k)`-ը․ հաշվե՛ք item-item cosine similarity-ն, վերցրե՛ք օգտատիրոջ ամենաբարձր գնահատականով առարկան և վերադարձրե՛ք k ամենանման այլ առարկաների ինդեքսները։

**Task:** Complete the function below so that the visible checks pass. / **Առաջադրանք՝** Լրացրե՛ք ֆունկցիան, որպեսզի տեսանելի թեստերը անցնեն։

**Q99 — your code / ձեր կոդը**

> Write your solution in the code cell below. / Գրե՛ք ձեր լուծումը ներքևի code cell-ում։

````python
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity

def recommend_items(matrix, user_idx, k):
    mat = np.array(matrix)
    
    item_similarity = cosine_similarity(mat.T)
    
    user_ratings = mat[user_idx]
    
    best_item_idx = np.argmax(user_ratings)
    
    sim_scores = item_similarity[best_item_idx]
    
    sorted_indices = np.argsort(sim_scores)[::-1]
    
    recommended_indices = [int(idx) for idx in sorted_indices if idx != best_item_idx][:k]
    
    return recommended_indices
````

**Q99 — visible checks (do not edit) / տեսանելի թեստեր (մի՛ փոփոխեք)**

````python
import numpy as np
matrix = np.array([[5, 0, 4, 1], [0, 5, 1, 4], [4, 1, 5, 0], [1, 4, 0, 5]], dtype=float)
recs = recommend_items(matrix, 2, 2)
assert isinstance(recs, list)
assert len(recs) == 2
assert 2 not in recs
assert all(0 <= i < matrix.shape[1] for i in recs)
print('visible checks passed', recs)
````
