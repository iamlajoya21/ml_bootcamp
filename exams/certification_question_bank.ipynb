### Q86 — Python — medium

**EN.** The function should return the last `n` items, but it is wrong. Give the corrected return statement.

**HY.** Ֆունկցիան պետք է վերադարձնի վերջին `n` տարրերը, բայց սխալ է։ Տվե՛ք ուղղված return-ը։

### Q94 — NP06 — reference solution

**Question:** Implement `standardize(x)` so it returns a NumPy array with mean 0 and population standard deviation 1. If the standard deviation is 0, return zeros.

**Rubric:** 2 marks correctness, 1 mark zero-std edge case, 1 mark vectorized NumPy use, 1 mark clarity.

````python
import numpy as np

def standardize(x):
    x = np.asarray(x, dtype=float)
    std = x.std()
    if std == 0:
        return np.zeros_like(x)
    return (x - x.mean()) / std
````

**Q94 — hidden tests**

````python
import numpy as np
x = np.array([1, 2, 3, 4], dtype=float) * 2 + 1
out = standardize(x)
assert out.shape == x.shape
assert np.allclose(out.mean(), 0.0)
assert np.allclose(out.std(), 1.0)
print('hidden checks passed')
````

### Q95 — PD11 — reference solution

**Question:** Implement `clean_and_summarize(df)`: copy the frame, parse `date` as datetime, fill missing `amount` with the mean of its `store`, create `adjusted` = `amount` * 2, and return the mean of `adjusted` per `store` sorted by group.

**Rubric:** 2 marks correct parsing, 2 marks grouped imputation, 1 mark new column, 1 mark grouped result, 1 mark no mutation of input.

````python
import pandas as pd

def clean_and_summarize(df):
    out = df.copy()
    out['date'] = pd.to_datetime(out['date'])
    out['amount'] = out.groupby('store')['amount'].transform(lambda s: s.fillna(s.mean()))
    out['adjusted'] = out['amount'] * 2
    return out.groupby('store')['adjusted'].mean().sort_index()
````

**Q95 — hidden tests**

````python
import pandas as pd
df = pd.DataFrame({'date': ['2024-02-01', '2024-02-02', '2024-02-03', '2024-02-04'], 'store': ['a', 'a', 'b', 'b'], 'amount': [2.0, None, 6.0, 8.0]})
out = clean_and_summarize(df)
assert abs(out.loc['a'] - (2.0 * 2)) < 1e-9
assert abs(out.loc['b'] - (7.0 * 2)) < 1e-9
print('hidden checks passed')
````

### Q96 — SK16 — reference solution

**Question:** Build and fit a scikit-learn Pipeline with median imputation, standard scaling, and a LogisticRegression. Return the fitted pipeline. The hidden tests require a validation score above 0.7.

**Rubric:** 2 marks pipeline structure, 2 marks correct preprocessing order, 2 marks fitted estimator, 1 mark no leakage.

````python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier

def build_model():
    model = LogisticRegression()
    return Pipeline([
        ('imputer', SimpleImputer(strategy='median')),
        ('scaler', StandardScaler()),
        ('model', model),
    ])
````

**Q96 — hidden tests**

````python
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
X, y = make_classification(n_samples=400, n_features=10, n_informative=6, random_state=1+100)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=1)
model = build_model()
model.fit(X_train, y_train)
score = accuracy_score(y_test, model.predict(X_test))
assert score >= 0.7, score
print('hidden checks passed', round(score, 3))
````

### Q97 — DL12 — reference solution

**Question:** Implement `SmallCNN(in_channels, num_classes)` with one Conv2d(16 filters, kernel 3, padding 1), ReLU, MaxPool2d(2), Flatten, and a Linear output layer. The hidden tests check the output shape for a (8, 3, 32, 32) input.

**Rubric:** 2 marks layer structure, 2 marks correct shapes, 1 mark ReLU/pooling, 1 mark clean forward pass.

````python
import torch
import torch.nn as nn

class SmallCNN(nn.Module):
    def __init__(self, in_channels, num_classes):
        super().__init__()
        self.features = nn.Sequential(
            nn.Conv2d(in_channels, 16, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2),
            nn.Flatten(),
            nn.Linear(16 * (32 // 2) * (32 // 2), num_classes),
        )

    def forward(self, x):
        return self.features(x)
````

**Q97 — hidden tests**

````python
import torch
x = torch.randn(2, 3, 32, 32)
model = SmallCNN(3, 10)
out = model(x)
assert out.shape == (2, 10), out.shape
print('hidden checks passed')
````

### Q98 — TS08 — reference solution

**Question:** Implement `forecast_next(series)` using lag-1 and lag-2 features with a linear regression. Return the one-step-ahead prediction as a float.

**Rubric:** 2 marks correct lag construction, 2 marks no leakage, 1 mark fitted regression, 1 mark float prediction.

````python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression

def forecast_next(series):
    s = pd.Series(series, dtype=float)
    X = pd.DataFrame({'lag1': s.shift(1), 'lag2': s.shift(2)}).dropna()
    y = s.iloc[2:]
    model = LinearRegression().fit(X, y)
    last = pd.DataFrame({'lag1': [s.iloc[-1]], 'lag2': [s.iloc[-2]]})
    return float(model.predict(last)[0])
````

**Q98 — hidden tests**

````python
import numpy as np
series = [x + 3 for x in [1, 2, 3, 4, 5, 6]]
pred = forecast_next(series)
assert isinstance(pred, float)
assert abs(pred - (7 + 3)) < 0.5, pred
print('hidden checks passed', round(pred, 3))
````

### Q99 — RC08 — reference solution

**Question:** Implement `recommend_items(matrix, user_idx, k)`: compute item-item cosine similarity, take the user's highest-rated item, and return the indices of the k most similar other items.

**Rubric:** 2 marks cosine similarity, 2 marks anchor selection, 1 mark exclusion, 1 mark top-k ordering.

````python
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity

def recommend_items(matrix, user_idx, k):
    X = np.asarray(matrix, dtype=float)
    sim = cosine_similarity(X.T)
    anchor = int(np.argmax(X[user_idx]))
    order = np.argsort(sim[anchor])[::-1]
    recs = [int(i) for i in order if int(i) != anchor][:k]
    return recs
````

**Q99 — hidden tests**

````python
import numpy as np
matrix = np.array([[5, 0, 4, 1], [0, 5, 1, 4], [4, 1, 5, 0], [1, 4, 0, 5]], dtype=float)
recs = recommend_items(matrix, 0, 2)
assert isinstance(recs, list)
assert len(recs) == 2
assert 0 not in recs
assert all(0 <= i < matrix.shape[1] for i in recs)
print('hidden checks passed', recs)
````

### Q100 — DP12 — reference solution

**Question:** Build a FastAPI app with `POST /predict` that accepts `{'x': float}` and returns `{'prediction': x * 2}`. The hidden tests use `fastapi.testclient.TestClient`.

**Rubric:** 2 marks endpoint, 2 marks Pydantic schema, 1 mark correct computation, 1 mark JSON response.

````python
from fastapi import FastAPI
from pydantic import BaseModel

app = FastAPI()

class Request(BaseModel):
    x: float

@app.post('/predict')
def predict(req: Request):
    return {'prediction': req.x * 2}
````

**Q100 — hidden tests**

````python
from fastapi.testclient import TestClient
client = TestClient(app)
resp = client.post('/predict', json={'x': 4.0})
assert resp.status_code == 200, resp.text
assert abs(resp.json()['prediction'] - 4.0 * 2) < 1e-9
print('hidden checks passed')
````

### Q94 — NP06 — reference solution

**Question:** Implement `standardize(x)` so it returns a NumPy array with mean 0 and population standard deviation 1. If the standard deviation is 0, return zeros.

**Rubric:** 2 marks correctness, 1 mark zero-std edge case, 1 mark vectorized NumPy use, 1 mark clarity.

````python
import numpy as np

def standardize(x):
    x = np.asarray(x, dtype=float)
    std = x.std()
    if std == 0:
        return np.zeros_like(x)
    return (x - x.mean()) / std
````

**Q94 — hidden tests**

````python
import numpy as np
x = np.array([10, 12, 14, 16, 18], dtype=float) * 2 + 1
out = standardize(x)
assert out.shape == x.shape
assert np.allclose(out.mean(), 0.0)
assert np.allclose(out.std(), 1.0)
print('hidden checks passed')
````

### Q95 — PD11 — reference solution

**Question:** Implement `clean_and_summarize(df)`: copy the frame, parse `day` as datetime, fill missing `sales` with the mean of its `region`, create `weighted` = `sales` * 1.5, and return the mean of `weighted` per `region` sorted by group.

**Rubric:** 2 marks correct parsing, 2 marks grouped imputation, 1 mark new column, 1 mark grouped result, 1 mark no mutation of input.

````python
import pandas as pd

def clean_and_summarize(df):
    out = df.copy()
    out['day'] = pd.to_datetime(out['day'])
    out['sales'] = out.groupby('region')['sales'].transform(lambda s: s.fillna(s.mean()))
    out['weighted'] = out['sales'] * 1.5
    return out.groupby('region')['weighted'].mean().sort_index()
````

**Q95 — hidden tests**

````python
import pandas as pd
df = pd.DataFrame({'day': ['2024-02-01', '2024-02-02', '2024-02-03', '2024-02-04'], 'region': ['a', 'a', 'b', 'b'], 'sales': [2.0, None, 6.0, 8.0]})
out = clean_and_summarize(df)
assert abs(out.loc['a'] - (2.0 * 1.5)) < 1e-9
assert abs(out.loc['b'] - (7.0 * 1.5)) < 1e-9
print('hidden checks passed')
````

### Q96 — SK16 — reference solution

**Question:** Build and fit a scikit-learn Pipeline with median imputation, standard scaling, and a RandomForestClassifier. Return the fitted pipeline. The hidden tests require a validation score above 0.7.

**Rubric:** 2 marks pipeline structure, 2 marks correct preprocessing order, 2 marks fitted estimator, 1 mark no leakage.

````python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier

def build_model():
    model = RandomForestClassifier()
    return Pipeline([
        ('imputer', SimpleImputer(strategy='median')),
        ('scaler', StandardScaler()),
        ('model', model),
    ])
````

### Q97 — DL12 — reference solution

**Question:** Implement `SmallCNN(in_channels, num_classes)` with one Conv2d(8 filters, kernel 3, padding 1), ReLU, MaxPool2d(2), Flatten, and a Linear output layer. The hidden tests check the output shape for a (4, 1, 28, 28) input.

**Rubric:** 2 marks layer structure, 2 marks correct shapes, 1 mark ReLU/pooling, 1 mark clean forward pass.

````python
import torch
import torch.nn as nn

class SmallCNN(nn.Module):
    def __init__(self, in_channels, num_classes):
        super().__init__()
        self.features = nn.Sequential(
            nn.Conv2d(in_channels, 8, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2),
            nn.Flatten(),
            nn.Linear(8 * (28 // 2) * (28 // 2), num_classes),
        )

    def forward(self, x):
        return self.features(x)
````

**Question:** Implement `forecast_next(series)` using lag-1 and lag-2 features with a linear regression. Return the one-step-ahead prediction as a float.

**Rubric:** 2 marks correct lag construction, 2 marks no leakage, 1 mark fitted regression, 1 mark float prediction.

````python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression

def forecast_next(series):
    s = pd.Series(series, dtype=float)
    X = pd.DataFrame({'lag1': s.shift(1), 'lag2': s.shift(2)}).dropna()
    y = s.iloc[2:]
    model = LinearRegression().fit(X, y)
    last = pd.DataFrame({'lag1': [s.iloc[-1]], 'lag2': [s.iloc[-2]]})
    return float(model.predict(last)[0])
````

**Q98 — hidden tests**

````python
import numpy as np
series = [x + 3 for x in [2, 4, 6, 8, 10]]
pred = forecast_next(series)
assert isinstance(pred, float)
assert abs(pred - (12 + 3)) < 0.5, pred
print('hidden checks passed', round(pred, 3))
````

### Q99 — RC08 — reference solution

**Question:** Implement `recommend_items(matrix, user_idx, k)`: compute item-item cosine similarity, take the user's highest-rated item, and return the indices of the k most similar other items.

**Rubric:** 2 marks cosine similarity, 2 marks anchor selection, 1 mark exclusion, 1 mark top-k ordering.

````python
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity

def recommend_items(matrix, user_idx, k):
    X = np.asarray(matrix, dtype=float)
    sim = cosine_similarity(X.T)
    anchor = int(np.argmax(X[user_idx]))
    order = np.argsort(sim[anchor])[::-1]
    recs = [int(i) for i in order if int(i) != anchor][:k]
    return recs
````

**Q99 — hidden tests**

````python
import numpy as np
matrix = np.array([[5, 0, 4, 1], [0, 5, 1, 4], [4, 1, 5, 0], [1, 4, 0, 5]], dtype=float)
recs = recommend_items(matrix, 0, 2)
assert isinstance(recs, list)
assert len(recs) == 2
assert 0 not in recs
assert all(0 <= i < matrix.shape[1] for i in recs)
print('hidden checks passed', recs)
````

### Q100 — DP12 — reference solution

**Question:** Build a FastAPI app with `POST /predict` that accepts `{'x': float}` and returns `{'prediction': x * 10}`. The hidden tests use `fastapi.testclient.TestClient`.

**Rubric:** 2 marks endpoint, 2 marks Pydantic schema, 1 mark correct computation, 1 mark JSON response.

````python
from fastapi import FastAPI
from pydantic import BaseModel

app = FastAPI()

class Request(BaseModel):
    x: float

@app.post('/predict')
def predict(req: Request):
    return {'prediction': req.x * 10}
````

**Q100 — hidden tests**

````python
from fastapi.testclient import TestClient
client = TestClient(app)
resp = client.post('/predict', json={'x': 4.0})
assert resp.status_code == 200, resp.text
assert abs(resp.json()['prediction'] - 4.0 * 10) < 1e-9
print('hidden checks passed')
````

### Q94 — NP06 — reference solution

**Question:** Implement `standardize(x)` so it returns a NumPy array with mean 0 and population standard deviation 1. If the standard deviation is 0, return zeros.

**Rubric:** 2 marks correctness, 1 mark zero-std edge case, 1 mark vectorized NumPy use, 1 mark clarity.

````python
import numpy as np

def standardize(x):
    x = np.asarray(x, dtype=float)
    std = x.std()
    if std == 0:
        return np.zeros_like(x)
    return (x - x.mean()) / std
````

**Q94 — hidden tests**

````python
import numpy as np
x = np.array([-2, 0, 2, 4], dtype=float) * 2 + 1
out = standardize(x)
assert out.shape == x.shape
assert np.allclose(out.mean(), 0.0)
assert np.allclose(out.std(), 1.0)
print('hidden checks passed')
````

### Q95 — PD11 — reference solution

**Question:** Implement `clean_and_summarize(df)`: copy the frame, parse `timestamp` as datetime, fill missing `score` with the mean of its `team`, create `scaled` = `score` * 10, and return the mean of `scaled` per `team` sorted by group.

**Rubric:** 2 marks correct parsing, 2 marks grouped imputation, 1 mark new column, 1 mark grouped result, 1 mark no mutation of input.

````python
import pandas as pd

def clean_and_summarize(df):
    out = df.copy()
    out['timestamp'] = pd.to_datetime(out['timestamp'])
    out['score'] = out.groupby('team')['score'].transform(lambda s: s.fillna(s.mean()))
    out['scaled'] = out['score'] * 10
    return out.groupby('team')['scaled'].mean().sort_index()
````

**Q95 — hidden tests**

````python
import pandas as pd
df = pd.DataFrame({'timestamp': ['2024-02-01', '2024-02-02', '2024-02-03', '2024-02-04'], 'team': ['a', 'a', 'b', 'b'], 'score': [2.0, None, 6.0, 8.0]})
out = clean_and_summarize(df)
assert abs(out.loc['a'] - (2.0 * 10)) < 1e-9
assert abs(out.loc['b'] - (7.0 * 10)) < 1e-9
print('hidden checks passed')
````

### Q96 — SK16 — reference solution

**Question:** Build and fit a scikit-learn Pipeline with median imputation, standard scaling, and a DecisionTreeClassifier. Return the fitted pipeline. The hidden tests require a validation score above 0.7.

**Rubric:** 2 marks pipeline structure, 2 marks correct preprocessing order, 2 marks fitted estimator, 1 mark no leakage.

````python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier

def build_model():
    model = DecisionTreeClassifier()
    return Pipeline([
        ('imputer', SimpleImputer(strategy='median')),
        ('scaler', StandardScaler()),
        ('model', model),
    ])
````

**Q96 — hidden tests**

````python
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
X, y = make_classification(n_samples=400, n_features=10, n_informative=6, random_state=3+100)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=1)
model = build_model()
model.fit(X_train, y_train)
score = accuracy_score(y_test, model.predict(X_test))
assert score >= 0.7, score
print('hidden checks passed', round(score, 3))
````

### Q97 — DL12 — reference solution

**Question:** Implement `SmallCNN(in_channels, num_classes)` with one Conv2d(32 filters, kernel 3, padding 1), ReLU, MaxPool2d(2), Flatten, and a Linear output layer. The hidden tests check the output shape for a (2, 3, 64, 64) input.

**Rubric:** 2 marks layer structure, 2 marks correct shapes, 1 mark ReLU/pooling, 1 mark clean forward pass.

````python
import torch
import torch.nn as nn

class SmallCNN(nn.Module):
    def __init__(self, in_channels, num_classes):
        super().__init__()
        self.features = nn.Sequential(
            nn.Conv2d(in_channels, 32, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2),
            nn.Flatten(),
            nn.Linear(32 * (64 // 2) * (64 // 2), num_classes),
        )

    def forward(self, x):
        return self.features(x)
````

**Q97 — hidden tests**

````python
import torch
x = torch.randn(2, 3, 64, 64)
model = SmallCNN(3, 100)
out = model(x)
assert out.shape == (2, 100), out.shape
print('hidden checks passed')
````

### Q98 — TS08 — reference solution

**Question:** Implement `forecast_next(series)` using lag-1 and lag-2 features with a linear regression. Return the one-step-ahead prediction as a float.

**Rubric:** 2 marks correct lag construction, 2 marks no leakage, 1 mark fitted regression, 1 mark float prediction.

````python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression

def forecast_next(series):
    s = pd.Series(series, dtype=float)
    X = pd.DataFrame({'lag1': s.shift(1), 'lag2': s.shift(2)}).dropna()
    y = s.iloc[2:]
    model = LinearRegression().fit(X, y)
    last = pd.DataFrame({'lag1': [s.iloc[-1]], 'lag2': [s.iloc[-2]]})
    return float(model.predict(last)[0])
````

**Q98 — hidden tests**

````python
import numpy as np
series = [x + 3 for x in [10, 20, 30, 40, 50]]
pred = forecast_next(series)
assert isinstance(pred, float)
assert abs(pred - (60 + 3)) < 0.5, pred
print('hidden checks passed', round(pred, 3))
````

### Q99 — RC08 — reference solution

**Question:** Implement `recommend_items(matrix, user_idx, k)`: compute item-item cosine similarity, take the user's highest-rated item, and return the indices of the k most similar other items.

**Rubric:** 2 marks cosine similarity, 2 marks anchor selection, 1 mark exclusion, 1 mark top-k ordering.

````python
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity

def recommend_items(matrix, user_idx, k):
    X = np.asarray(matrix, dtype=float)
    sim = cosine_similarity(X.T)
    anchor = int(np.argmax(X[user_idx]))
    order = np.argsort(sim[anchor])[::-1]
    recs = [int(i) for i in order if int(i) != anchor][:k]
    return recs
````

**Q99 — hidden tests**

````python
import numpy as np
matrix = np.array([[5, 0, 4, 1], [0, 5, 1, 4], [4, 1, 5, 0], [1, 4, 0, 5]], dtype=float)
recs = recommend_items(matrix, 0, 2)
assert isinstance(recs, list)
assert len(recs) == 2
assert 0 not in recs
assert all(0 <= i < matrix.shape[1] for i in recs)
print('hidden checks passed', recs)
````

### Q100 — DP12 — reference solution

**Question:** Build a FastAPI app with `POST /predict` that accepts `{'x': float}` and returns `{'prediction': x * 0.5}`. The hidden tests use `fastapi.testclient.TestClient`.

**Rubric:** 2 marks endpoint, 2 marks Pydantic schema, 1 mark correct computation, 1 mark JSON response.

````python
from fastapi import FastAPI
from pydantic import BaseModel

app = FastAPI()

class Request(BaseModel):
    x: float

@app.post('/predict')
def predict(req: Request):
    return {'prediction': req.x * 0.5}
````

**Q100 — hidden tests**

````python
from fastapi.testclient import TestClient
client = TestClient(app)
resp = client.post('/predict', json={'x': 4.0})
assert resp.status_code == 200, resp.text
assert abs(resp.json()['prediction'] - 4.0 * 0.5) < 1e-9
print('hidden checks passed')
````

### Q94 — NP06 — reference solution

**Question:** Implement `standardize(x)` so it returns a NumPy array with mean 0 and population standard deviation 1. If the standard deviation is 0, return zeros.

**Rubric:** 2 marks correctness, 1 mark zero-std edge case, 1 mark vectorized NumPy use, 1 mark clarity.

````python
import numpy as np

def standardize(x):
    x = np.asarray(x, dtype=float)
    std = x.std()
    if std == 0:
        return np.zeros_like(x)
    return (x - x.mean()) / std
````

**Q94 — hidden tests**

````python
import numpy as np
x = np.array([5, 5.5, 6, 6.5, 7], dtype=float) * 2 + 1
out = standardize(x)
assert out.shape == x.shape
assert np.allclose(out.mean(), 0.0)
assert np.allclose(out.std(), 1.0)
print('hidden checks passed')
````

### Q95 — PD11 — reference solution

**Question:** Implement `clean_and_summarize(df)`: copy the frame, parse `event_date` as datetime, fill missing `price` with the mean of its `category`, create `taxed` = `price` * 1.2, and return the mean of `taxed` per `category` sorted by group.

**Rubric:** 2 marks correct parsing, 2 marks grouped imputation, 1 mark new column, 1 mark grouped result, 1 mark no mutation of input.

````python
import pandas as pd

def clean_and_summarize(df):
    out = df.copy()
    out['event_date'] = pd.to_datetime(out['event_date'])
    out['price'] = out.groupby('category')['price'].transform(lambda s: s.fillna(s.mean()))
    out['taxed'] = out['price'] * 1.2
    return out.groupby('category')['taxed'].mean().sort_index()
````

````python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier

def build_model():
    model = KNeighborsClassifier()
    return Pipeline([
        ('imputer', SimpleImputer(strategy='median')),
        ('scaler', StandardScaler()),
        ('model', model),
    ])
````

### Q97 — DL12 — reference solution

**Question:** Implement `SmallCNN(in_channels, num_classes)` with one Conv2d(12 filters, kernel 3, padding 1), ReLU, MaxPool2d(2), Flatten, and a Linear output layer. The hidden tests check the output shape for a (16, 4, 16, 16) input.

**Rubric:** 2 marks layer structure, 2 marks correct shapes, 1 mark ReLU/pooling, 1 mark clean forward pass.

````python
import torch
import torch.nn as nn

class SmallCNN(nn.Module):
    def __init__(self, in_channels, num_classes):
        super().__init__()
        self.features = nn.Sequential(
            nn.Conv2d(in_channels, 12, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2),
            nn.Flatten(),
            nn.Linear(12 * (16 // 2) * (16 // 2), num_classes),
        )

    def forward(self, x):
        return self.features(x)
````

**Question:** Implement `forecast_next(series)` using lag-1 and lag-2 features with a linear regression. Return the one-step-ahead prediction as a float.

**Rubric:** 2 marks correct lag construction, 2 marks no leakage, 1 mark fitted regression, 1 mark float prediction.

````python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression

def forecast_next(series):
    s = pd.Series(series, dtype=float)
    X = pd.DataFrame({'lag1': s.shift(1), 'lag2': s.shift(2)}).dropna()
    y = s.iloc[2:]
    model = LinearRegression().fit(X, y)
    last = pd.DataFrame({'lag1': [s.iloc[-1]], 'lag2': [s.iloc[-2]]})
    return float(model.predict(last)[0])
````

**Q98 — hidden tests**

````python
import numpy as np
series = [x + 3 for x in [5, 8, 11, 14, 17]]
pred = forecast_next(series)
assert isinstance(pred, float)
assert abs(pred - (20 + 3)) < 0.5, pred
print('hidden checks passed', round(pred, 3))
````

### Q99 — RC08 — reference solution

**Question:** Implement `recommend_items(matrix, user_idx, k)`: compute item-item cosine similarity, take the user's highest-rated item, and return the indices of the k most similar other items.

**Rubric:** 2 marks cosine similarity, 2 marks anchor selection, 1 mark exclusion, 1 mark top-k ordering.

````python
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity

def recommend_items(matrix, user_idx, k):
    X = np.asarray(matrix, dtype=float)
    sim = cosine_similarity(X.T)
    anchor = int(np.argmax(X[user_idx]))
    order = np.argsort(sim[anchor])[::-1]
    recs = [int(i) for i in order if int(i) != anchor][:k]
    return recs
````

**Q99 — hidden tests**

````python
import numpy as np
matrix = np.array([[5, 0, 4, 1], [0, 5, 1, 4], [4, 1, 5, 0], [1, 4, 0, 5]], dtype=float)
recs = recommend_items(matrix, 0, 2)
assert isinstance(recs, list)
assert len(recs) == 2
assert 0 not in recs
assert all(0 <= i < matrix.shape[1] for i in recs)
print('hidden checks passed', recs)
````

### Q100 — DP12 — reference solution

**Question:** Build a FastAPI app with `POST /predict` that accepts `{'x': float}` and returns `{'prediction': x * 3}`. The hidden tests use `fastapi.testclient.TestClient`.

**Rubric:** 2 marks endpoint, 2 marks Pydantic schema, 1 mark correct computation, 1 mark JSON response.

````python
from fastapi import FastAPI
from pydantic import BaseModel

app = FastAPI()

class Request(BaseModel):
    x: float

@app.post('/predict')
def predict(req: Request):
    return {'prediction': req.x * 3}
````

**Q100 — hidden tests**

````python
from fastapi.testclient import TestClient
client = TestClient(app)
resp = client.post('/predict', json={'x': 4.0})
assert resp.status_code == 200, resp.text
assert abs(resp.json()['prediction'] - 4.0 * 3) < 1e-9
print('hidden checks passed')
````
